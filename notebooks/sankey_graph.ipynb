```python
df_cluster1 = df_cluster.copy()

df_cluster1 = df_cluster1[df_cluster1["type_of_sample"] == "Stool"]
```

```python
df_cluster1["node"] = df_cluster1["time_of_sample"] + "_C" + df_cluster1["cluster"].astype("str")
len_cluster = df_cluster1["node"].unique()
index_for_cluster = dict(zip(len_cluster, range(len(len_cluster))))
df_cluster1["index_for_cluster"] = df_cluster1["node"].replace(index_for_cluster)
```

```python
df_cluster1['person_id'] = df_cluster1['sample_id'].str.split('-').str[0] + '-' + df_cluster1['sample_id'].str.split('-').str[1]

df_t1 = df_cluster1[df_cluster1['time_of_sample'] == 'T1']
df_t2 = df_cluster1[df_cluster1['time_of_sample'] == 'T2']
df_t3 = df_cluster1[df_cluster1['time_of_sample'] == 'T3']

merged_t1_t2 = pd.merge(df_t1, df_t2, on='person_id', suffixes=('_T1', '_T2'))

links_t1_t2 = merged_t1_t2.groupby(['index_for_cluster_T1', 'index_for_cluster_T2']).size().reset_index(name='value')
links_t1_t2 = links_t1_t2.rename(columns={'index_for_cluster_T1': 'source', 'index_for_cluster_T2': 'target'})

merged_t2_t3 = pd.merge(df_t2, df_t3, on='person_id', suffixes=('_T2', '_T3'))

links_t2_t3 = merged_t2_t3.groupby(['index_for_cluster_T2', 'index_for_cluster_T3']).size().reset_index(name='value')
links_t2_t3 = links_t2_t3.rename(columns={'index_for_cluster_T2': 'source', 'index_for_cluster_T3': 'target'})

all_links = pd.concat([links_t1_t2, links_t2_t3], ignore_index=True)
```

**Figura 1:** *Figura Periodo de transición para Eces*

```python
node_data = df_cluster1[['node', 'index_for_cluster']].drop_duplicates().sort_values('index_for_cluster')
node_labels = node_data['node'].tolist()

link_sources = all_links['source'].tolist()
link_targets = all_links['target'].tolist()
link_values = all_links['value'].tolist()

color_cluster = {
    # Clúster 1 (Azules)
    'T1_C1': 'rgba(66, 135, 245, 0.4)',   # azul claro
    'T2_C1': 'rgba(30, 90, 210, 0.6)',    # azul medio
    'T3_C1': 'rgba(12, 55, 140, 0.9)',    # azul intenso

    # Clúster 2 (Naranjas / Coral)
    'T1_C2': 'rgba(255, 159, 64, 0.4)',   # naranja claro
    'T2_C2': 'rgba(255, 120, 50, 0.6)',   # naranja medio
    'T3_C2': 'rgba(210, 85, 35, 0.9)',    # naranja intenso

    # Clúster 3 (Verdes)
    'T1_C3': 'rgba(75, 192, 192, 0.4)',   # verde claro
    'T2_C3': 'rgba(40, 160, 160, 0.6)',   # verde medio
    'T3_C3': 'rgba(5, 120, 120, 0.9)',    # verde intenso
}

node_colors = node_data['node'].map(color_cluster).fillna('rgba(201, 203, 207, 0.6)').tolist()

color_map = {
    'T1_C1': 'rgba(66, 135, 245, 0.4)',   # azul claro
    'T2_C1': 'rgba(30, 90, 210, 0.6)',    # azul medio
    'T3_C1': 'rgba(12, 55, 140, 0.9)',    # azul intenso

    # Clúster 2 (Naranjas / Coral)
    'T1_C2': 'rgba(255, 159, 64, 0.4)',   # naranja claro
    'T2_C2': 'rgba(255, 120, 50, 0.6)',   # naranja medio
    'T3_C2': 'rgba(210, 85, 35, 0.9)',    # naranja intenso
}

all_links['link_category'] = all_links['source'].map(node_data.set_index('index_for_cluster')['node'])
link_colors = all_links['link_category'].map(color_map).fillna('rgba(201, 203, 207, 0.6)').tolist()

fig = go.Figure(data=[go.Sankey(

    node = dict(
      pad = 15,
      thickness = 20,
      line = dict(color = "black", width = 0.5),
      label = node_labels, 
      color = node_colors
    ),

    link = dict(
      source = link_sources,
      target = link_targets,
      value = link_values,
      color = link_colors
    )
)])

fig.update_layout(
    title_text="Diagrama de Sankey: Flujo de Clústers (T1 a T3)",
    font_size=12
)

fig.show()
```

```python
df_cluster3 = df_cluster.copy()
df_cluster3 = df_cluster3[df_cluster3["type_of_sample"] == "Milk"]

df_cluster3["node"] = df_cluster3["time_of_sample"] + "_C" + df_cluster3["cluster"].astype("str")
len_cluster = df_cluster3["node"].unique()
index_for_cluster = dict(zip(len_cluster, range(len(len_cluster))))
df_cluster3["index_for_cluster"] = df_cluster3["node"].replace(index_for_cluster)

df_cluster3['person_id'] = df_cluster3['sample_id'].str.split('-').str[0] + '-' + df_cluster3['sample_id'].str.split('-').str[1]

node_data = df_cluster3[['node', 'index_for_cluster']].drop_duplicates().sort_values('index_for_cluster')
node_labels = df_cluster3['node'].tolist()

all_people = sorted(df_cluster3['person_id'].unique())
dropdown_options = [{'label': 'Mostrar Todos', 'value': 'All'}] + [{'label': person, 'value': person} for person in all_people]

df_t1 = df_cluster3[df_cluster3['time_of_sample'] == 'T1']
df_t2 = df_cluster3[df_cluster3['time_of_sample'] == 'T2']
df_t3 = df_cluster3[df_cluster3['time_of_sample'] == 'T3']

merged_t1_t2 = pd.merge(df_t1, df_t2, on='person_id', suffixes=('_T1', '_T2'))
merged_t2_t3 = pd.merge(df_t2, df_t3, on='person_id', suffixes=('_T2', '_T3'))
```

Heces - Gráfico de trasabilidad

```python
df_cluster4 = df_cluster.copy()
df_cluster4 = df_cluster4[df_cluster4["type_of_sample"] == "Stool"]

# Definición de Nodos
df_cluster4["node"] = df_cluster4["time_of_sample"] + "_C" + df_cluster4["cluster"].astype("str")
len_cluster = sorted(df_cluster4["node"].unique()) # Ordenar para consistencia
index_for_cluster = dict(zip(len_cluster, range(len(len_cluster))))
df_cluster4["index_for_cluster"] = df_cluster4["node"].replace(index_for_cluster)

# Extracción de person_id
df_cluster4['person_id'] = df_cluster4['sample_id'].str.split('-').str[0] + '-' + df_cluster4['sample_id'].str.split('-').str[1]

# Preparación de datos de nodos (esto es estático)
node_data = df_cluster4[['node', 'index_for_cluster']].drop_duplicates().sort_values('index_for_cluster')
node_labels = node_data['node'].tolist()

# Obtener lista de personas para el Dropdown
all_people = sorted(df_cluster4['person_id'].unique())
dropdown_options = [{'label': 'Mostrar Todos', 'value': 'All'}] + [{'label': person, 'value': person} for person in all_people]

# --- DATOS PARA LOS ENLACES (ANTES DE AGREGAR) ---
df_t1 = df_cluster4[df_cluster4['time_of_sample'] == 'T1']
df_t2 = df_cluster4[df_cluster4['time_of_sample'] == 'T2']
df_t3 = df_cluster4[df_cluster4['time_of_sample'] == 'T3']

# Datos individuales (con person_id)
merged_t1_t2_individual = pd.merge(df_t1, df_t2, on='person_id', suffixes=('_T1', '_T2'))
merged_t2_t3_individual = pd.merge(df_t2, df_t3, on='person_id', suffixes=('_T2', '_T3'))

# Datos agregados (los totales, tu lógica original)
links_t1_t2_agg = merged_t1_t2_individual.groupby(['index_for_cluster_T1', 'index_for_cluster_T2']).size().reset_index(name='value')
links_t1_t2_agg = links_t1_t2_agg.rename(columns={'index_for_cluster_T1': 'source', 'index_for_cluster_T2': 'target'})

links_t2_t3_agg = merged_t2_t3_individual.groupby(['index_for_cluster_T2', 'index_for_cluster_T3']).size().reset_index(name='value')
links_t2_t3_agg = links_t2_t3_agg.rename(columns={'index_for_cluster_T2': 'source', 'index_for_cluster_T3': 'target'})

# Este es el total de flujos
all_links_agg = pd.concat([links_t1_t2_agg, links_t2_t3_agg], ignore_index=True)

app = dash.Dash(__name__)

app.layout = html.Div([
    html.H1("Trazabilidad de Clústers (Resaltando Persona)"),
    
    html.Label("Selecciona una Persona para Resaltar:"),
    dcc.Dropdown(
        id='person-dropdown',
        options=dropdown_options,
        value='All' 
    ),
    
    dcc.Graph(id='sankey-graph')
])

@app.callback(
    Output('sankey-graph', 'figure'),
    Input('person-dropdown', 'value')
)
def update_sankey(selected_person_id):

    highlight_color = '#E35408'
    default_color = 'rgba(200, 200, 200, 0.5)'

    if selected_person_id == 'All':

        link_sources = all_links_agg['source'].tolist()
        link_targets = all_links_agg['target'].tolist()
        link_values = all_links_agg['value'].tolist()
        link_colors = ['#CADAF1'] * len(link_sources)
        title_text = "Diagrama de Sankey: Flujo de Todos (T1 a T3)"
    
    else:
        title_text = f"Diagrama de Sankey: Resaltando a {selected_person_id}"

        person_path_t1 = merged_t1_t2_individual[merged_t1_t2_individual['person_id'] == selected_person_id][['index_for_cluster_T1', 'index_for_cluster_T2']]
        person_path_t1.columns = ['source', 'target']
        
        person_path_t2 = merged_t2_t3_individual[merged_t2_t3_individual['person_id'] == selected_person_id][['index_for_cluster_T2', 'index_for_cluster_T3']]
        person_path_t2.columns = ['source', 'target']
        
        person_full_path = pd.concat([person_path_t1, person_path_t2])

        new_sources = []
        new_targets = []
        new_values = []
        new_colors = []

        for _, row in all_links_agg.iterrows():
            source = row['source']
            target = row['target']
            total_value = row['value']

            person_took_path = ((person_full_path['source'] == source) & 
                                (person_full_path['target'] == target)).any()

            if person_took_path:

                new_sources.append(source)
                new_targets.append(target)
                new_values.append(1)
                new_colors.append(highlight_color)

                if total_value > 1:
                    new_sources.append(source)
                    new_targets.append(target)
                    new_values.append(total_value - 1)
                    new_colors.append(default_color)
            
            else:
                new_sources.append(source)
                new_targets.append(target)
                new_values.append(total_value)
                new_colors.append(default_color)
        
        link_sources = new_sources
        link_targets = new_targets
        link_values = new_values
        link_colors = new_colors

    fig = go.Figure(data=[go.Sankey(
        node = dict(
          pad = 15,
          thickness = 20,
          line = dict(color = "black", width = 0.5),
          label = node_labels,
          color = "#43799F"
        ),
        
        link = dict(
          source = link_sources,
          target = link_targets,
          value = link_values,
          color = link_colors
        )
    )])

    fig.update_layout(
        title_text=title_text,
        font_size=12
    )

    return fig

app.run(debug=True)
```
